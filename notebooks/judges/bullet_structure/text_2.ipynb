```python
TEXT_TO_JUDGE = "../../../data/text_02_long_text.txt"
```

```python
with open(TEXT_TO_JUDGE, "r") as f:
    text_to_judge = f.read()
```

```python
from promptsmith.dspy_init import get_dspy
dspy, lm = get_dspy()
```

```python
from promptsmith.judges.judge_bullet_structure import JudgeBulletStructure
judge = dspy.Predict(JudgeBulletStructure)
```

just for fun, let's see how the judge handles the text before it's even bulletized:

```python
r1 = judge(text=text_to_judge)

print("Reasoning:", r1.reasoning)
print("Score:", r1.score)
```

```python
from promptsmith.tasks.bad_to_good.bulletize_text_1 import BulletizeText as simple_bulletizer
from promptsmith.tasks.bad_to_good.bulletize_text_2 import BulletizeText as better_bulletizer

simple_bulletizer = dspy.Predict(simple_bulletizer)
better_bulletizer = dspy.Predict(better_bulletizer)
```

```python
bulletized_text_1_simple = simple_bulletizer(input_text=text_to_judge)
```

```python
r2 = judge(text=bulletized_text_1_simple)

print("Reasoning:", r2.reasoning)
print("Score:", r2.score)
```

```python
bulletized_text_1_better = better_bulletizer(input_text=text_to_judge)
```

```python
r3 = judge(text=bulletized_text_1_better)

print("Reasoning:", r3.reasoning)
print("Score:", r3.score)
```

# Interview with Dorna Moini, CEO of Documate

Dorna Moini discusses her journey from litigator to legal tech entrepreneur and the evolution of Documate.

## Background
- **Dorna Moini:** Founder and CEO of Documate, a document automation platform.
- **Previous Experience:** Litigator at Sheppard Mullin and Sidley Austin.
- **Recognition:** Named a Legal Rebel by the American Bar Association in 2019.
- **Teaching Role:** Will teach Legal Technology at USC Law School starting January 2020.
- Dorna's career path shifted unexpectedly due to pro bono work with domestic violence survivors.

---

## Domestic Violence Platform
- **Pro Bono Work:** Dorna spent significant hours helping domestic violence survivors.
- **Automation Idea:** Wanted to create a tool similar to TurboTax for legal documents.
- **Collaboration:** Partnered with Michael, her co-founder, to build the platform.
- **Launch:** The platform was used by legal aid organizations in California.
- **Current Use:** Although not available for consumers, legal aid organizations still utilize it.

---

## Transition to Documate
- **Reason for Transition:** The need for scalable solutions across jurisdictions.
- **Platform Goals:** To allow lawyers to automate their expertise and collaborate on tools.
- **Comparison to GitHub:** Aims to create a collaborative environment for legal professionals.

---

## Document Automation vs. Expert Systems
- **Definitions:** Document automation involves decision trees; expert systems can be more complex.
- **No Code Features:** Designed for users without coding knowledge to build workflows.
- **Advanced Tools:** Some clients integrate complex databases for enhanced functionality.

---

## Future Vision for Documate
- **2020 Goals:** 
  - Become the most powerful document automation software.
  - Provide affordable access for small and mid-sized law firms.
  - Enhance data functionality for client information management.
  - Foster collaboration among lawyers through shared workflows.

---

## Best Practices for Document Automation
- **Start with Familiar Problems:** Focus on issues you have personally experienced.
- **Stakeholder Engagement:** Consult with potential users to understand their needs.
- **Document Automation Specialists:** A new group to assist with best practices and user testing.

---

## Law School Curriculum Updates
- **Innovative Thinking:** Encourage students to think outside traditional legal frameworks.

## Regulatory Changes
