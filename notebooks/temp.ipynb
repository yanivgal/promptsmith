```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Read the data
df = pd.read_csv('../data/documents.csv')

# Randomly select 1000 examples for test
test_indices = np.random.choice(len(df), size=1000, replace=False)
test_df = df.iloc[test_indices]
train_df = df.drop(test_indices)

# Save the splits
test_df.to_csv('../data/documents_test.csv', index=False)
train_df.to_csv('../data/documents_train.csv', index=False)

print(f"Total examples: {len(df)}")
print(f"Test set size: {len(test_df)}")
print(f"Training set size: {len(train_df)}")
```
